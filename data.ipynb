```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

df = pd.read_csv("dataset.csv")

top_artists = df['artists'].value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(
    x=top_artists.values, 
    y=top_artists.index, 
    hue=top_artists.index,
    palette='coolwarm', 
    dodge=False,
    legend=False
)
plt.title("Top 10 Artists by Number of Tracks")
plt.xlabel("Number of Tracks")
plt.ylabel("Artist")
plt.tight_layout()
plt.savefig("images/top_artists.png")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['popularity'], bins=30, kde=True, color='skyblue')
plt.title("Distribution of Track Popularity")
plt.xlabel("Popularity Score")
plt.ylabel("Number of Tracks")
plt.tight_layout()
plt.savefig("images/popularity_distribution.png")
plt.show()
```

```python
df['duration_sec'] = df['duration_ms'] / 1000

filtered = df[(df['duration_sec'] >= 30) & (df['duration_sec'] <= 600)]

plt.figure(figsize=(10, 5))
sns.histplot(filtered['duration_sec'], bins=40, kde=True, color='orange')
plt.title("Song Duration Distribution (30s–10min)")
plt.xlabel("Duration (seconds)")
plt.ylabel("Number of Tracks")
plt.tight_layout()
plt.savefig("images/duration_distribution_filtered.png")
plt.show()
```

```python
features = ['popularity', 'duration_ms', 'tempo', 'liveness', 'valence', 'energy', 'danceability']

plt.figure(figsize=(10, 8))
sns.heatmap(df[features].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Between Audio Features")
plt.tight_layout()
plt.savefig("images/feature_correlation.png")
plt.show()
```
